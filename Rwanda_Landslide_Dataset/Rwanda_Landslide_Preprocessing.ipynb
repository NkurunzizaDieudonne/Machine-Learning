```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Rwanda_Landslide_Dataset.csv")

print("Dataset loaded successfully.")

df.head()

df.info()
df.describe()
missing_values = df.isnull().sum()

print(missing_values)

missing_columns = df.isnull().sum()

missing_columns = missing_columns[missing_columns > 0]

print("Features containing missing values:")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])


missing_columns = df.isnull().sum()
missing_columns = missing_columns[missing_columns > 0]

print("Features containing missing values:")
print(missing_columns)
plt.figure(figsize=(12, 6))

sns.heatmap(df.isnull(), cbar=False)

plt.title("Missing Values Heatmap")
plt.show()

plt.figure(figsize=(6, 4))

sns.countplot(x="landslide_occurred", data=df)

plt.title("Landslide Occurrence Distribution")
plt.xlabel("Landslide Occurred")
plt.ylabel("Number of Records")
plt.show()
plt.figure(figsize=(14, 10))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()



print("Missing values before filling:")
print(df.isnull().sum())


numeric_columns = df.select_dtypes(include="number").columns


for column in numeric_columns:
    if column != "landslide_occurred":
        df[column] = df[column].fillna(df[column].mean())


categorical_columns = df.select_dtypes(include="object").columns


for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])


print("\nMissing values after filling:")
print(df.isnull().sum())
import matplotlib.pyplot as plt


numeric_features = [col for col in numeric_columns 
                    if col != "landslide_occurred"]


outlier_counts = {}

for column in numeric_features:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_limit) | 
                  (df[column] > upper_limit)]

    outlier_counts[column] = len(outliers)


print("Number of outliers in each numerical feature:")
for column, count in outlier_counts.items():
    print(column, ":", count)


total_outliers = sum(outlier_counts.values())

print("\nTotal number of outliers:", total_outliers)

target = "landslide_occurred"


num_features = df.select_dtypes(include="number").columns.tolist()
num_features.remove(target)


cat_features = df.select_dtypes(include="object").columns.tolist()

print("Numerical features:")
print(num_features)

print("\nCategorical features:")
print(cat_features)
from sklearn.model_selection import train_test_split


X = df.drop("landslide_occurred", axis=1)
y = df["landslide_occurred"]


X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OrdinalEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_features),
        ("cat", OrdinalEncoder(
            handle_unknown="use_encoded_value",
            unknown_value=-1
        ), cat_features)
    ]
)

X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed = preprocessor.transform(X_test)

print("Transformed training data:", X_train_transformed.shape)
print("Transformed testing data:", X_test_transformed.shape)
```
